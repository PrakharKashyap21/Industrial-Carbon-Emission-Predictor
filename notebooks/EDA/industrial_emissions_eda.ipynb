# Industrial Carbon Emission Prediction System — Exploratory Data Analysis (Phase 3)

This notebook performs exploratory data analysis (EDA), statistical summary, distribution analysis, correlation studies, and feature engineering validation for the **Industrial Carbon Emission Prediction System** dataset.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure Matplotlib style
plt.style.use("seaborn-v0_8-darkgrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11
```

## 1. Load Processed Industrial Dataset

```python
dataset_path = "../../data/processed/industrial_emissions_processed.csv"
df = pd.read_csv(dataset_path)
df["timestamp"] = pd.to_datetime(df["timestamp"])
print(f"Dataset Shape: {df.shape}")
df.head()
```

## 2. Descriptive Statistics

```python
df.describe().T[["mean", "std", "min", "50%", "max"]]
```

## 3. Correlation Analysis with CO₂ Emission

```python
num_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(12, 10))
sns.heatmap(num_df.corr(), annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

## 4. Key Relationships vs Target (actual_co2_emission_kg)

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.scatterplot(data=df, x="electricity_consumption_kwh", y="actual_co2_emission_kg", hue="plant_id", ax=axes[0, 0], palette="viridis")
axes[0, 0].set_title("Electricity Consumption vs CO₂ Emission")
axes[0, 0].set_xlabel("Electricity Consumption (kWh)")
axes[0, 0].set_ylabel("CO₂ Emission (kg)")

sns.scatterplot(data=df, x="diesel_consumption_liters", y="actual_co2_emission_kg", hue="plant_id", ax=axes[0, 1], palette="viridis")
axes[0, 1].set_title("Diesel Consumption vs CO₂ Emission")
axes[0, 1].set_xlabel("Diesel Consumption (Liters)")
axes[0, 1].set_ylabel("CO₂ Emission (kg)")

sns.scatterplot(data=df, x="natural_gas_consumption_m3", y="actual_co2_emission_kg", hue="plant_id", ax=axes[1, 0], palette="viridis")
axes[1, 0].set_title("Natural Gas Consumption vs CO₂ Emission")
axes[1, 0].set_xlabel("Natural Gas Consumption (m³)")
axes[1, 0].set_ylabel("CO₂ Emission (kg)")

sns.scatterplot(data=df, x="production_quantity", y="actual_co2_emission_kg", hue="plant_id", ax=axes[1, 1], palette="viridis")
axes[1, 1].set_title("Production Quantity vs CO₂ Emission")
axes[1, 1].set_xlabel("Production Quantity (Units)")
axes[1, 1].set_ylabel("CO₂ Emission (kg)")

plt.tight_layout()
plt.show()
```

## 5. Plant-Level Emissions Summary

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="plant_id", y="actual_co2_emission_kg", palette="Set2")
plt.title("CO₂ Emission Distribution Across Industrial Facilities")
plt.xlabel("Plant ID")
plt.ylabel("CO₂ Emission (kg)")
plt.show()
```
